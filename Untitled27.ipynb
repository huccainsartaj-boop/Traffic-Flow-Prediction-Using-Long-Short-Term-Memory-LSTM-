```python
import os
import glob
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import kagglehub

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices("GPU"))
```

```python
path = kagglehub.dataset_download(
    "fedesoriano/traffic-prediction-dataset"
)

print("Dataset folder:", path)

csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

print("CSV files found:", csv_files)

if not csv_files:
    raise FileNotFoundError(
        "CSV file nahi mili. Dataset path check karein."
    )

# Is dataset mein traffic.csv expected hai
preferred = [
    f for f in csv_files
    if os.path.basename(f).lower() == "traffic.csv"
]

CSV_PATH = preferred[0] if preferred else csv_files[0]

print("Selected CSV:", CSV_PATH)
```

```python
df = pd.read_csv(CSV_PATH)

# Column names clean karein
df.columns = df.columns.str.strip()

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all"))
```

```python
required_columns = {"DateTime", "Junction", "Vehicles"}

missing_columns = required_columns - set(df.columns)

if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )

df["DateTime"] = pd.to_datetime(
    df["DateTime"],
    errors="coerce"
)

df["Junction"] = pd.to_numeric(
    df["Junction"],
    errors="coerce"
)

df["Vehicles"] = pd.to_numeric(
    df["Vehicles"],
    errors="coerce"
)

df = df.dropna(
    subset=["DateTime", "Junction", "Vehicles"]
)

df = df[df["Vehicles"] >= 0].copy()

df["Junction"] = df["Junction"].astype(int)

# Same junction aur same timestamp ke duplicate records
df = df.drop_duplicates(
    subset=["DateTime", "Junction"],
    keep="last"
)

df = df.sort_values(
    ["Junction", "DateTime"]
).reset_index(drop=True)

print("Cleaned shape:", df.shape)
print("\nJunctions:", sorted(df["Junction"].unique()))
display(df.head())
```

```python
plt.figure(figsize=(14, 5))

for junction in sorted(df["Junction"].unique()):
    temp = df[df["Junction"] == junction]

    plt.plot(
        temp["DateTime"],
        temp["Vehicles"],
        label=f"Junction {junction}",
        linewidth=1
    )

plt.title("Traffic Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Vehicles")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
junction_summary = (
    df.groupby("Junction")
    .agg(
        observations=("Vehicles", "size"),
        start_date=("DateTime", "min"),
        end_date=("DateTime", "max"),
        average_vehicles=("Vehicles", "mean"),
        max_vehicles=("Vehicles", "max")
    )
)

display(junction_summary)
```

```python
df["Hour"] = df["DateTime"].dt.hour
df["DayOfWeek"] = df["DateTime"].dt.dayofweek
df["Month"] = df["DateTime"].dt.month

# Cyclical encoding: 23:00 aur 00:00 ek doosre ke qareeb hain
df["Hour_sin"] = np.sin(2 * np.pi * df["Hour"] / 24)
df["Hour_cos"] = np.cos(2 * np.pi * df["Hour"] / 24)

df["Day_sin"] = np.sin(
    2 * np.pi * df["DayOfWeek"] / 7
)
df["Day_cos"] = np.cos(
    2 * np.pi * df["DayOfWeek"] / 7
)

print(df[
    ["DateTime", "Junction", "Vehicles",
     "Hour_sin", "Hour_cos", "Day_sin", "Day_cos"]
].head())
```

```python
LOOKBACK = 24

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

FEATURES = [
    "Vehicles",
    "Hour_sin",
    "Hour_cos",
    "Day_sin",
    "Day_cos"
]

TARGET_INDEX = FEATURES.index("Vehicles")

print("Past time steps:", LOOKBACK)
print("Input features:", FEATURES)
print("Target: next time step Vehicles")
```

```python
train_parts = []
val_parts = []
test_parts = []

junction_scalers = {}

for junction in sorted(df["Junction"].unique()):

    group = (
        df[df["Junction"] == junction]
        .sort_values("DateTime")
        .copy()
        .reset_index(drop=True)
    )

    # Is dataset mein hourly data hai. Agar timestamps
    # missing hon to artificial continuous sequence na banayein.
    if len(group) < LOOKBACK + 20:
        print(
            f"Skipping Junction {junction}: "
            "not enough observations"
        )
        continue

    # Missing hours identify karein
    hour_diff = group["DateTime"].diff().dropna()
    print(
        f"Junction {junction}: "
        f"{len(group)} rows, "
        f"{(hour_diff != pd.Timedelta(hours=1)).sum()} "
        "non-hourly gaps"
    )

    values = group[FEATURES].to_numpy(dtype=np.float32)

    n = len(values)
    train_end = int(n * TRAIN_RATIO)
    val_end = int(n * (TRAIN_RATIO + VAL_RATIO))

    if train_end <= LOOKBACK:
        print("Not enough training rows.")
        continue

    # Scaler only sees training observations
    scaler = StandardScaler()
    scaler.fit(values[:train_end])

    scaled = scaler.transform(values).astype(np.float32)
    junction_scalers[junction] = scaler

    def create_sequences(data, start, end, lookback):
        X, y = [], []

        for target_pos in range(start, end):
            if target_pos < lookback:
                continue

            X.append(data[target_pos-lookback:target_pos])
            y.append(data[target_pos, TARGET_INDEX])

        if not X:
            return (
                np.empty((0, lookback, data.shape[1]),
                         dtype=np.float32),
                np.empty((0,), dtype=np.float32)
            )

        return np.array(X, dtype=np.float32), np.array(
            y, dtype=np.float32
        )

    # Training targets: training period
    Xtr, ytr = create_sequences(
        scaled, LOOKBACK, train_end, LOOKBACK
    )

    # Validation targets: validation period
    # Past training observations are allowed as input context
    Xv, yv = create_sequences(
        scaled, train_end, val_end, LOOKBACK
    )

    # Test targets: final unseen period
    Xt, yt = create_sequences(
        scaled, val_end, n, LOOKBACK
    )

    if len(Xtr):
        train_parts.append((Xtr, ytr))

    if len(Xv):
        val_parts.append((Xv, yv))

    if len(Xt):
        test_parts.append((Xt, yt))


def combine_parts(parts, lookback, n_features):
    if not parts:
        return (
            np.empty((0, lookback, n_features), dtype=np.float32),
            np.empty((0,), dtype=np.float32)
        )

    X = np.concatenate([p[0] for p in parts], axis=0)
    y = np.concatenate([p[1] for p in parts], axis=0)
    return X, y


X_train, y_train = combine_parts(
    train_parts, LOOKBACK, len(FEATURES)
)

X_val, y_val = combine_parts(
    val_parts, LOOKBACK, len(FEATURES)
)

X_test, y_test = combine_parts(
    test_parts, LOOKBACK, len(FEATURES)
)

print("\nTraining:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)

if min(len(X_train), len(X_val), len(X_test)) == 0:
    raise ValueError(
        "Kisi split mein sequences nahi banin. "
        "CSV timestamps aur junction sizes check karein."
    )
```

```python
train_parts = []
val_parts = []
test_parts = []
junction_scalers = {}

for junction in sorted(df["Junction"].unique()):

    group = (
        df[df["Junction"] == junction]
        .sort_values("DateTime")
        .reset_index(drop=True)
    )

    values = group[FEATURES].to_numpy(dtype=np.float32)
    times = group["DateTime"].to_numpy()

    n = len(group)
    train_end = int(n * TRAIN_RATIO)
    val_end = int(n * (TRAIN_RATIO + VAL_RATIO))

    if train_end <= LOOKBACK or val_end <= train_end:
        continue

    scaler = StandardScaler()
    scaler.fit(values[:train_end])
    scaled = scaler.transform(values).astype(np.float32)

    junction_scalers[junction] = scaler

    # Consecutive one-hour windows only
    Xtr, ytr, Xv, yv, Xt, yt = [], [], [], [], [], []

    for target_pos in range(LOOKBACK, n):

        window_times = times[target_pos-LOOKBACK:target_pos+1]
        diffs = np.diff(window_times)

        # Exactly 1-hour intervals throughout the input
        # and between the last input and target
        if not np.all(
            diffs == np.timedelta64(1, "h")
        ):
            continue

        x_window = scaled[target_pos-LOOKBACK:target_pos]
        target = scaled[target_pos, TARGET_INDEX]

        if target_pos < train_end:
            Xtr.append(x_window)
            ytr.append(target)

        elif target_pos < val_end:
            Xv.append(x_window)
            yv.append(target)

        else:
            Xt.append(x_window)
            yt.append(target)

    def pack(X, y):
        if len(X) == 0:
            return (
                np.empty(
                    (0, LOOKBACK, len(FEATURES)),
                    dtype=np.float32
                ),
                np.empty((0,), dtype=np.float32)
            )
        return np.asarray(X, dtype=np.float32), np.asarray(
            y, dtype=np.float32
        )

    a, b = pack(Xtr, ytr)
    c, d = pack(Xv, yv)
    e, f = pack(Xt, yt)

    if len(a):
        train_parts.append((a, b))
    if len(c):
        val_parts.append((c, d))
    if len(e):
        test_parts.append((e, f))

X_train, y_train = combine_parts(
    train_parts, LOOKBACK, len(FEATURES)
)
X_val, y_val = combine_parts(
    val_parts, LOOKBACK, len(FEATURES)
)
X_test, y_test = combine_parts(
    test_parts, LOOKBACK, len(FEATURES)
)

print("Gap-safe training:", X_train.shape, y_train.shape)
print("Gap-safe validation:", X_val.shape, y_val.shape)
print("Gap-safe testing:", X_test.shape, y_test.shape)

if min(len(X_train), len(X_val), len(X_test)) == 0:
    raise ValueError(
        "Gap-safe sequences kam hain. "
        "LOOKBACK ko 12 karein ya timestamp gaps inspect karein."
    )
```

```python
tf.keras.backend.clear_session()

model = Sequential([
    Input(shape=(LOOKBACK, len(FEATURES))),

    LSTM(
        64,
        return_sequences=True
    ),
    Dropout(0.20),

    LSTM(32),
    Dropout(0.20),

    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=1e-6,
    verbose=1
)

checkpoint = ModelCheckpoint(
    "/content/best_traffic_lstm.keras",
    monitor="val_loss",
    save_best_only=True,
    verbose=1
)

callbacks = [
    early_stop,
    reduce_lr,
    checkpoint
]
```

```python
EPOCHS = 40
BATCH_SIZE = 64

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=False,
    callbacks=callbacks,
    verbose=1
)
```

```python
MODEL_PATH = "/content/best_traffic_lstm.keras"

if os.path.exists(MODEL_PATH):
    model = load_model(MODEL_PATH)
    print("Best model loaded:", MODEL_PATH)
else:
    print("Checkpoint nahi mila; current model use hoga.")
```

```python
results = []
prediction_frames = []

for junction in sorted(df["Junction"].unique()):

    group = (
        df[df["Junction"] == junction]
        .sort_values("DateTime")
        .reset_index(drop=True)
    )

    if junction not in junction_scalers:
        continue

    scaler = junction_scalers[junction]
    values = group[FEATURES].to_numpy(dtype=np.float32)

    n = len(values)
    train_end = int(n * TRAIN_RATIO)
    val_end = int(n * (TRAIN_RATIO + VAL_RATIO))

    scaled = scaler.transform(values).astype(np.float32)

    Xj, yj, dates_j = [], [], []

    times = group["DateTime"].to_numpy()

    for target_pos in range(max(LOOKBACK, val_end), n):

        window_times = times[target_pos-LOOKBACK:target_pos+1]

        if not np.all(
            np.diff(window_times) == np.timedelta64(1, "h")
        ):
            continue

        Xj.append(scaled[target_pos-LOOKBACK:target_pos])
        yj.append(values[target_pos, TARGET_INDEX])
        dates_j.append(group.loc[target_pos, "DateTime"])

    if not Xj:
        continue

    Xj = np.asarray(Xj, dtype=np.float32)
    y_true = np.asarray(yj, dtype=np.float32)
    y_pred_scaled = model.predict(Xj, verbose=0).reshape(-1)

    # Convert scaled predictions to the original vehicle scale
    y_pred = (
        y_pred_scaled * scaler.scale_[TARGET_INDEX]
        + scaler.mean_[TARGET_INDEX]
    )

    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred) if len(y_true) > 1 else np.nan

    results.append({
        "Junction": junction,
        "Test Samples": len(y_true),
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    prediction_frames.append(pd.DataFrame({
        "DateTime": dates_j,
        "Junction": junction,
        "Actual": y_true,
        "Predicted": y_pred
    }))

results_df = pd.DataFrame(results)

display(results_df.round(3))

if prediction_frames:
    predictions_df = pd.concat(
        prediction_frames,
        ignore_index=True
    )
    print("Prediction rows:", len(predictions_df))
else:
    raise ValueError(
        "Test predictions nahi banin. "
        "Timestamp gaps aur split sizes check karein."
    )
```
